### Task 2 - Missing Values, Leakage, and Outliers

```python
# Count missing values in each column
missing_count = data.isnull().sum()

# Calculate missing-value percentage for each column
missing_percentage = (missing_count / len(data)) * 100

# Create a table showing count and percentage
missing_values = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

# Display only columns that have missing values
display(missing_values[missing_values["Missing Count"] > 0])
```

```python
# df will be the cleaned dataset
df = data.drop(columns=['company'])
```

The company column contains mostly missing values, so it provides very little useful information for the analysis. Therefore, removing this column is a reasonable choice.

```python
# Remove reservation status columns since they are not needed for the analysis
df = df.drop(columns=['reservation_status', 'reservation_status_date'])
```

```python
# Select numerical columns to check for outliers
outlier_columns = ["lead_time", "adr", "adults", "children", "babies"]

# Create the figure
plt.figure(figsize=(12, 8))

# Create a boxplot for each selected column
for i, column in enumerate(outlier_columns, 1):
    plt.subplot(2, 3, i)

    # Create the boxplot
    box = plt.boxplot(
        df[column].dropna(),
        patch_artist=True
    )

    # Give each box a different color
    colors = ["skyblue", "lightgreen", "lightcoral", "plum", "khaki"]
    box["boxes"][0].set_facecolor(colors[i - 1])

    # Add title and label
    plt.title(column, fontsize=12, fontweight="bold")
    plt.ylabel("Value")

    # Add a light grid to make the plot easier to read
    plt.grid(axis="y", linestyle="--", alpha=0.4)

# Adjust the spacing
plt.tight_layout()

# Display the plots
plt.show()
```

```python
# Calculate the first and third quartiles of the 'adr' column
Q1 = df['adr'].quantile(0.25)
Q3 = df['adr'].quantile(0.75)

# Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1

# Store the initial number of rows
initial = len(df)

# Define the lower and upper limits for detecting outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Keep only the rows where 'adr' falls within the IQR limits
df = df[(df['adr'] >= lower_bound) & (df['adr'] <= upper_bound)]

# Calculate how many rows were removed as outliers
rows_removed = initial - len(df)

# Display the number of removed outlier rows
print(f"Removed {rows_removed} outlier rows based on the 'adr' column IQR bounds.")
```

Observation : Boxplots were used to inspect selected numerical features for possible outliers. The IQR method was then used on `lead_time` and `adr` to identify clear extreme values. Only observations outside the IQR limits were removed. The number of removed rows is reported in the output above.
