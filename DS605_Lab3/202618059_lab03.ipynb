```python
# Select numerical columns to check for outliers
outlier_columns = ["lead_time", "adr", "adults", "children", "babies"]

# Create the figure
plt.figure(figsize=(12, 8))

# Create a boxplot for each selected column
for i, column in enumerate(outlier_columns, 1):
    plt.subplot(2, 3, i)

    # Create the boxplot
    box = plt.boxplot(
        df[column].dropna(),
        patch_artist=True
    )

    # Give each box a different color
    colors = ["skyblue", "lightgreen", "lightcoral", "plum", "khaki"]
    box["boxes"][0].set_facecolor(colors[i - 1])

    # Add title and label
    plt.title(column, fontsize=12, fontweight="bold")
    plt.ylabel("Value")

    # Add a light grid to make the plot easier to read
    plt.grid(axis="y", linestyle="--", alpha=0.4)

# Adjust the spacing
plt.tight_layout()

# Display the plots
plt.show()
```

```python
# Calculate the first and third quartiles of the 'adr' column
Q1 = df['adr'].quantile(0.25)
Q3 = df['adr'].quantile(0.75)

# Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1

# Store the initial number of rows
initial = len(df)

# Define the lower and upper limits for detecting outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Keep only the rows where 'adr' falls within the IQR limits
df = df[(df['adr'] >= lower_bound) & (df['adr'] <= upper_bound)]

# Calculate how many rows were removed as outliers
rows_removed = initial - len(df)

# Display the number of removed outlier rows
print(f"Removed {rows_removed} outlier rows based on the 'adr' column IQR bounds.")
```

Observation : Boxplots were used to inspect selected numerical features for possible outliers. The IQR method was then used on `lead_time` and `adr` to identify clear extreme values. Only observations outside the IQR limits were removed. The number of removed rows is reported in the output above.

### Task 3 - Create Two Preprocessing Pipelines

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder
```

```python
y = df['is_canceled']
X = df.drop('is_canceled', axis=1)
```

```python
# Split the data using train_test_split(test_size=0.2, stratify=y, random_state=42). Use the same split for all experiments.
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
num_cols = X_train.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = X_train.select_dtypes(include=['object', 'category']).columns.tolist()
```

```python
cat_pipe = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy="most_frequent")),
    ('encoder', OneHotEncoder(handle_unknown="ignore"))
])

# standarisation
num_pipe_a = Pipeline(steps=[
    ('imputer', KNNImputer(n_neighbors=5)),
    ('scaler', StandardScaler())
])

# min maz normalization
num_pipe_b = Pipeline(steps=[
    ('imputer', KNNImputer(n_neighbors=5)),
    ('scaler', MinMaxScaler())
])

preprocessor_a = ColumnTransformer(transformers=[
    ('num', num_pipe_a, num_cols),
    ('cat', cat_pipe, cat_cols)
])

preprocessor_b = ColumnTransformer(transformers=[
    ('num', num_pipe_b, num_cols),
    ('cat', cat_pipe, cat_cols)
])
```

## Part B - Model Training and Performance Evaluation
-----

### Task 4 - Train Two Classification Models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline
```

```python
log_reg = LogisticRegression(max_iter=1000)
tree_clf = DecisionTreeClassifier(random_state=42)
```

```python
pipeline_logreg_a = Pipeline(steps=[
    ('preprocessor', preprocessor_a),
    ('classifier', log_reg)
])

pipeline_logreg_b = Pipeline(steps=[
    ('preprocessor', preprocessor_b),
    ('classifier', log_reg)
])
```

```python
pipeline_tree_a = Pipeline(steps=[
    ('preprocessor', preprocessor_a),
    ('classifier', tree_clf)
])

pipeline_tree_b = Pipeline(steps=[
    ('preprocessor', preprocessor_b),
    ('classifier', tree_clf)
])
```

```python
print("Logistic Regression with Pipeline A (StandardScaler)")
pipeline_logreg_a.fit(X_train, y_train)

print("Logistic Regression with Pipeline B (MinMaxScaler)")
pipeline_logreg_b.fit(X_train, y_train)

print("Decision Tree with Pipeline A (StandardScaler)")
pipeline_tree_a.fit(X_train, y_train)

print("Decision Tree with Pipeline B (MinMaxScaler)")
pipeline_tree_b.fit(X_train, y_train)
```

### Task 5 - Evaluate and Compare the Four Results

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, ConfusionMatrixDisplay
)
```

```python
def evaluate_model(model, name, X_train, y_train, X_test, y_test):
    
    # Get predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    return {
        'Model Configuration': name,
        'Train Accuracy': accuracy_score(y_train, y_train_pred),
        'Test Accuracy': accuracy_score(y_test, y_test_pred),
        'Precision': precision_score(y_test, y_test_pred),
        'Recall': recall_score(y_test, y_test_pred),
        'F1-Score': f1_score(y_test, y_test_pred)
    }
```

```python
results = []
results.append(evaluate_model(pipeline_logreg_a, "LogReg + StandardScaler", X_train, y_train, X_test, y_test))
results.append(evaluate_model(pipeline_logreg_b, "LogReg + MinMaxScaler", X_train, y_train, X_test, y_test))
results.append(evaluate_model(pipeline_tree_a, "Decision Tree + StandardScaler", X_train, y_train, X_test, y_test))
results.append(evaluate_model(pipeline_tree_b, "Decision Tree + MinMaxScaler", X_train, y_train, X_test, y_test))

results_df = pd.DataFrame(results)
print(results_df)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

cm_lr = confusion_matrix(y_test, pipeline_logreg_a.predict(X_test))
disp_lr = ConfusionMatrixDisplay(confusion_matrix=cm_lr, display_labels=['Not Canceled', 'Canceled'])
disp_lr.plot(ax=axes[0], cmap='Blues', values_format='d')
axes[0].set_title('Best Logistic Regression\n(StandardScaler)')

cm_dt = confusion_matrix(y_test, pipeline_tree_a.predict(X_test))
disp_dt = ConfusionMatrixDisplay(confusion_matrix=cm_dt, display_labels=['Not Canceled', 'Canceled'])
disp_dt.plot(ax=axes[1], cmap='Greens', values_format='d')
axes[1].set_title('Best Decision Tree\n(StandardScaler)')

plt.tight_layout()
plt.show()
```

## OBSERVATIONS

1. **Best Overall Model:**  
   Decision Tree + MinMaxScaler gives the best overall performance, with a Test Accuracy of approximately **0.8595** and an F1-Score of **0.8114**. It slightly outperforms Decision Tree + StandardScaler and performs better than both Logistic Regression models.

2. **Effect of Scaling on Logistic Regression:**  
   StandardScaler performs slightly better than MinMaxScaler for Logistic Regression. The Test Accuracy is **0.8154 vs. 0.8121**, while the F1-Score is **0.7274 vs. 0.7208**. Thus, StandardScaler is more suitable for Logistic Regression in this case.

3. **Effect of Scaling on Decision Tree:**  
   Scaling has almost no effect on the Decision Tree. The Test Accuracy is **0.8592 with StandardScaler** and **0.8595 with MinMaxScaler**. This is because Decision Trees are generally not sensitive to the scale of input features.

4. **Overfitting:**  
   Logistic Regression shows a small difference between training and testing accuracy, indicating good generalization. In comparison, the Decision Tree has a train-test gap of approximately **13.7%**, which indicates significant overfitting.

5. **Precision and Recall:**  
   The Decision Tree achieves a higher Recall of approximately **0.815**, compared with around **0.65–0.66** for Logistic Regression. This means the Decision Tree is better at identifying customers who are likely to cancel their bookings.

6. **Overall Conclusion:**  
   The **Decision Tree + MinMaxScaler** is the best-performing model based on the evaluation metrics. However, its higher train-test gap indicates overfitting. Therefore, techniques such as limiting the tree depth or increasing the minimum samples per leaf could be used to improve its generalization.
