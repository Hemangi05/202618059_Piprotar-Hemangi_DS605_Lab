### Task 2 - Vectorized Arithmetic and Linear Algebra

Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @ or np.matmul().

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nMatrix Addition:")
print(A + B)

print("\nElement-wise Multiplication:")
print(A * B)

print("\nMatrix Multiplication:")
print(A @ B)
```

For a square matrix, find its transpose and determinant.

```python
print("Transpose of A:")
print(A.T)

print("\nDeterminant of A:")
print(np.linalg.det(A))
```

If invertible, find the inverse and verify A @ A^-1 using np.allclose().

```python
det_A = np.linalg.det(A)

if not np.isclose(det_A, 0):
    A_inverse = np.linalg.inv(A)

    print("Inverse of A:")
    print(A_inverse)

    print("\nA @ A^-1:")
    print(A @ A_inverse)

    print("\nVerification:")
    print(np.allclose(A @ A_inverse, np.eye(2)))
else:
    print("Matrix A is not invertible.")
```

### Task 5 — Filtering and Querying

```python
# Problem 1: Male passengers older than 50: how many?

male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]

print("Male passengers older than 50:", len(male_over_50))
```

```python
# Problem 2: Female first-class passengers: how many,
# and what percentage survived?

female_first_class = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]

count = len(female_first_class)
survival_percentage = female_first_class["Survived"].mean() * 100

print("Female first-class passengers:", count)
print("Percentage survived:", survival_percentage)
```

```python
# Problem 3: Age 20-40, Fare above overall median,
# and survived: how many?

median_fare = df["Fare"].median()

result = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > median_fare) &
    (df["Survived"] == 1)
]

print("Overall median Fare:", median_fare)
print("Number of passengers:", len(result))
```

```python
# Problem 4: Travelling alone (SibSp=0, Parch=0),
# age < 30, and did not survive: how many?

result = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("Number of passengers:", len(result))
```

```python
# Problem 5: Embarked='S', Pclass 2 or 3,
# and Fare above Southampton median: how many?

southampton_median = df[df["Embarked"] == "S"]["Fare"].median()

result = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median)
]

print("Southampton median Fare:", southampton_median)
print("Number of passengers:", len(result))
```

## Task 6 - groupby() and Aggregation

```python
# Problem 1: Find the survival rate by Sex.

survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("Survival Rate by Sex:")
print(survival_by_sex)
```

```python
# Problem 2: Find the survival rate by Pclass.

survival_by_pclass = df.groupby("Pclass")["Survived"].mean() * 100

print("Survival Rate by Pclass:")
print(survival_by_pclass)
```

```python
# Problem 3: Find the average Age and Fare by Pclass.

average_by_pclass = df.groupby("Pclass")[["Age", "Fare"]].mean()

print("Average Age and Fare by Pclass:")
print(average_by_pclass)
```

```python
# Problem 4: Find the passenger count and survival rate by Sex-Pclass.

result = df.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

result["Survival_Rate"] = result["Survival_Rate"] * 100

print("Passenger Count and Survival Rate by Sex-Pclass:")
print(result)
```

```python
# Problem 5: Find the passenger count, average Fare,
# and survival rate by Embarked.

result = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

result["Survival_Rate"] = result["Survival_Rate"] * 100

print("Passenger Count, Average Fare, and Survival Rate by Embarked:")
print(result)
```

### Task 7 — Missing Values and Fare Outliers

```python
import matplotlib.pyplot as plt
```

```python
# Problem 1: Find the missing-value count and percentage for every column.
# Plot missing counts as a bar chart.

missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

print("Missing Values Summary:")
display(missing_summary)

missing_count.plot(
    kind="bar",
    figsize=(10, 5),
    title="Missing Values by Column"
)

plt.xlabel("Columns")
plt.ylabel("Missing Count")
plt.show()
```

```python
# Problem 2: Fill missing Age values with the mean Age.
# Show missing counts before and after.

print("Missing Age values before filling:")
print(df["Age"].isnull().sum())

mean_age = df["Age"].mean()

df["Age"] = df["Age"].fillna(mean_age)

print("\nMean Age:", mean_age)

print("\nMissing Age values after filling:")
print(df["Age"].isnull().sum())
```

```python
# Problem 3: Try to fill NaN values by imputing using:
# mean, median, mode, and random value.

import numpy as np
import pandas as pd

# Load the original dataset again
df_original = pd.read_csv("train.csv")

# Mean imputation
df_mean = df_original.copy()
df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())

# Median imputation
df_median = df_original.copy()
df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())

# Mode imputation
df_mode = df_original.copy()
df_mode["Age"] = df_mode["Age"].fillna(df_mode["Age"].mode()[0])

# Random value imputation
df_random = df_original.copy()

missing_indices = df_random["Age"].isnull()

random_values = np.random.choice(
    df_random["Age"].dropna(),
    size=missing_indices.sum()
)

df_random.loc[missing_indices, "Age"] = random_values

print("Missing Age values after Mean imputation:",
      df_mean["Age"].isnull().sum())

print("Missing Age values after Median imputation:",
      df_median["Age"].isnull().sum())

print("Missing Age values after Mode imputation:",
      df_mode["Age"].isnull().sum())

print("Missing Age values after Random imputation:",
      df_random["Age"].isnull().sum())
```

```python
# Problem 4: For Fare, calculate Q1, Q3, IQR,
# 1.5xIQR bounds, and count the outliers.

Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare Outliers:", len(outliers))
```

### Task 8 - Features and Pivot Table

```python
# Problem 1: Create FamilySize = SibSp + Parch + 1
# and IsAlone = 1 when FamilySize = 1.

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

print("FamilySize and IsAlone:")
display(df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head(10))
```

```python
# Problem 2: Create a pivot table with rows=Sex,
# columns=Pclass, values=mean Survived.
# Find the highest and lowest groups.

pivot_table = pd.pivot_table(
    df,
    index="Sex",
    columns="Pclass",
    values="Survived",
    aggfunc="mean"
)

print("Survival Rate Pivot Table:")
display(pivot_table)

# Find highest and lowest survival rate
highest_group = pivot_table.stack().idxmax()
lowest_group = pivot_table.stack().idxmin()

highest_value = pivot_table.stack().max()
lowest_value = pivot_table.stack().min()

print("Highest survival group:", highest_group)
print("Highest survival rate:", highest_value)

print("\nLowest survival group:", lowest_group)
print("Lowest survival rate:", lowest_value)
```

### Task 9 - Visualizations and Observations

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Problem 1: Plot a correlation heatmap for relevant numerical columns.
# Note the strongest positive and negative relationships.

numerical_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]

correlation = df[numerical_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()

# Find strongest positive and negative relationships
corr_pairs = correlation.where(
    np.triu(np.ones(correlation.shape), k=1).astype(bool)
).stack()

print("Strongest positive relationship:")
print(corr_pairs.idxmax(), "=", corr_pairs.max())

print("\nStrongest negative relationship:")
print(corr_pairs.idxmin(), "=", corr_pairs.min())
```
